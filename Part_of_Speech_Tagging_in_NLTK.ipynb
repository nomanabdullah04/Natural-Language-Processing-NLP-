```python
import nltk
nltk.download('punkt_tab')
sentences=nltk.sent_tokenize(paragraph)
nltk.download('stopwords')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
#apply for find the pos tag
for i in range(len(sentences)):
  words=nltk.word_tokenize(sentences[i])
  words=[word for word in words if word.lower() not in nltk.corpus.stopwords.words('english')]
  tagged_words=nltk.pos_tag(words)
  print(tagged_words)

```

```python
#another example

parm="Taj Mahal is a beautiful Mounment"

from nltk.tokenize import word_tokenize
words=word_tokenize(parm)
tagged_words=nltk.pos_tag(words)
print(tagged_words)
```

### Example for Named Entity Recognition (NER)

Let's use the following sentence for Named Entity Recognition:

```python
nltk.download('maxent_ne_chunker')
nltk.download('words')
nltk.download('maxent_ne_chunker_tab')
```

```python
sentence_ner = "Google was founded by Larry Page and Sergey Brin in California."

from nltk.tokenize import word_tokenize
words = word_tokenize(sentence_ner)
tagged_words = nltk.pos_tag(words)
#ne_chunk function used for determine this is noue for other
namedEnt = nltk.ne_chunk(tagged_words)
print(namedEnt)
```

```python
nltk.ne_chunk(tagged_words).draw()
```
